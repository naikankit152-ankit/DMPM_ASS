```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [100, 150, 120, 180, 200]
products = ['A', 'B', 'C', 'D']
product_sales = [250, 180, 300, 220]
marks = [45, 55, 60, 62, 65, 70, 72, 75, 80, 85, 90, 95]
hours = [1, 2, 3, 4, 5, 6, 7, 8]
scores = [40, 45, 50, 55, 65, 70, 80, 90]
```

```python
plt.bar(products, product_sales, color='skyblue')

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
plt.plot(months, sales, marker='o', color='blue')

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```

```python
plt.pie(
    product_sales,
    labels=products,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Product Sales Distribution")
plt.show()
```

```python
plt.hist(marks, bins=5, color='orange', edgecolor='black')

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.scatter(hours, scores, color='red')

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.grid(True)
plt.show()
```

```python
plt.boxplot(marks)

plt.title("Marks Distribution")
plt.ylabel("Marks")

plt.show()
```

```python
plt.fill_between(months, sales, color='lightgreen', alpha=0.7)

plt.plot(months, sales, color='green')

plt.title("Monthly Sales - Area Plot")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```
