```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = np.random.randint(10, 100, 10)
df = pd.DataFrame(data, columns=["Value"])
print("Original Dataset:")
print(df)
```

```python
minmax_scaler = MinMaxScaler()
df["Min-Max"] = minmax_scaler.fit_transform(df[["Value"]])
print("\nMin-Max Normalization:")
print(df[["Value", "Min-Max"]])
```

```python
zscore_scaler = StandardScaler()
df["Z-Score"] = zscore_scaler.fit_transform(df[["Value"]])
print("\nZ-Score Normalization:")
print(df[["Value", "Z-Score"]])
```

```python
max_value = df["Value"].abs().max()
j = len(str(int(max_value)))
df["Decimal Scaling"] = df["Value"] / (10 ** j)
print("\nDecimal Scaling Normalization:")
print(df[["Value", "Decimal Scaling"]])
```

```python
print("\nFinal Dataset:")
print(df)
```

```python
import pandas as pd
from scipy.stats import chi2_contingency
data = {
    "Product_A": [50, 30],
    "Product_B": [10, 40],
    }
df = pd.DataFrame(data, index=["Men", "Women"])
print("--- Observed Contingency Table ---")
print(df)
print("\n" + "=" * 40 + "\n")
stat, p_value, dof, expected = chi2_contingency(df)
print(f"Chi-Square Statistic : {stat:.4f}")
print(f"P-Value              : {p_value:.4f}")
print(f"Degrees of Freedom   : {dof}")
print("\n--- Expected Frequencies Table ---")
print(pd.DataFrame(expected, index=df.index, columns=df.columns).round(2))

print("\n" + "=" * 40 + "\n")
alpha = 0.05
if p_value <= alpha:
    print("Result: SIGNIFICANT (Reject H0). There is a significant association between the variables.")
else:
    print("Result: NOT SIGNIFICANT (Fail to reject H0). The variables appear to be independent.")
```
