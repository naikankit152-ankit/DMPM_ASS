```python
#mode
data=[10,20,20,30,30,40,50,50]
counts={}
for ele in data:
  if ele in counts:
    counts[ele]+=1
  else:
    counts[ele]=1
freq=max(counts.values())
mode=[ele for ele,frq in counts.items() if frq==freq]
print(f"mode={mode}")
```

```python
def calculate_mean(data_list):
    if not data_list:
        return 0
    return sum(data_list) / len(data_list)
```

```python
def calculate_median(data_list):
    if not data_list:
        raise ValueError("Cannot calculate median of an empty list.")
    sorted_data = sorted(data_list)
    n = len(sorted_data)
    if n % 2 == 0:
        mid1 = sorted_data[n // 2 - 1]
        mid2 = sorted_data[n // 2]
        median = (mid1 + mid2) / 2
    else:
        median = sorted_data[n // 2]
    return median
```

```python
def calculate_mode(data_list):
    if not data_list:
        return []
    counts = {}
    for ele in data_list:
        counts[ele] = counts.get(ele, 0) + 1
    if not counts:
        return []
    max_freq = max(counts.values())
    modes = [ele for ele, frq in counts.items() if frq == max_freq]
    return modes
```

```python
def calculate_variance(data_list):
    if not data_list or len(data_list) == 0:
        raise ValueError("Cannot calculate variance of an empty list.")
    mean = calculate_mean(data_list)
    squared_differences_sum = sum([(x - mean)**2 for x in data_list])
    variance = squared_differences_sum / len(data_list)
    return variance
```

```python
import math
def calculate_std_dev(data_list):
    if not data_list or len(data_list) == 0:
        raise ValueError("Cannot calculate standard deviation of an empty list.")
    variance = calculate_variance(data_list)
    std_dev = math.sqrt(variance)
    return std_dev
```

```python
demo_data = [10, 20, 20, 30, 30, 40, 50, 50, 50, 60]
print(f"Original Data: {demo_data}")
mean_val = calculate_mean(demo_data)
print(f"Mean: {mean_val}")
median_val = calculate_median(demo_data)
print(f"Median: {median_val}")
mode_val = calculate_mode(demo_data)
print(f"Mode(s): {mode_val}")
variance_val = calculate_variance(demo_data)
print(f"Variance (Population): {variance_val}")
std_dev_val = calculate_std_dev(demo_data)
print(f"Standard Deviation (Population): {std_dev_val}")
```

```python
data_for_variance_sd = [10, 20, 30, 40, 50, 50]
print(f"Data for Variance and Standard Deviation: {data_for_variance_sd}")
variance_result = calculate_variance(data_for_variance_sd)
print(f"Variance: {variance_result}")
std_dev_result = calculate_std_dev(data_for_variance_sd)
print(f"Standard Deviation: {std_dev_result}")
```
